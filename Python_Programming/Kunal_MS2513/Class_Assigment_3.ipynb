Problem 2. Customer Age Analysis
Using the DataFrame, determine:
• Average age of the customers
• Minimum age
• Maximum age
• Number of male customers
• Number of female customers

```python
# 1) Average age of the customers 
print("Average age:", df["Age"].mean() )
```

```python
#  2) Minimum age
print("Min Age:",df["Age"].min())
```

```python
# 3) Maximum age
print("Maximum age:",df["Age"].max())
```

```python
# 4) No. of Male Customers 
print("No. of male customer: ",(df["Gender"] == "Male").sum())
```

```python
# 5) No. of Female Customers 
print("No. of Female customer: ",(df["Gender"] == "Female").sum())
```
