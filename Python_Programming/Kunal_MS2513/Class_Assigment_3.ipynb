Problem 7. Create a Bank Account Class
Create a class called BankAccount.
The class should have the following attributes:
• account num
• owner
• balance
Use init () to initialize these attributes.
Create two different bank account objects and display their attributes.

```python
# Create a class 
class BankAccount:
# Define a init method
    def __init__(self, account_num, owner, balance):
               self.account_num = account_num
               self.owner= owner
               self.balance = balance

# define a Deposite method here 
    def deposite(self, amount):
         self.balance += amount

# Define a Withrawal method 
    def withdraw(self, amount):
        if self.balance >= amount:
            self.balance -= amount
            return True
        else:
            return False

# Define Str method 
    def __str__(self):
        return f"{self.account_num} | {self.owner}: ₹{self.balance}"
```

```python
account1 = BankAccount(1,'Kunal', 1000)

print(account1.account_num)
print(account1.owner)
print(account1.balance)
```

```python
account2 = BankAccount(2,'Pankaj', 500)

print(account2.account_num)
print(account2.owner)
print(account2.balance)
```

Problem 11. Student Class
Create a class called Student.
The class should have the following attributes:
• name
• roll no
• marks
Create at least three Student objects.
Add a method called average marks() that returns the average of the student’s marks.
Display the average marks of each student.

```python
# class of student 
class Student :
    def __init__(self, name, roll_no, marks):
        self.name = name
        self.roll_no = roll_no
        self.marks = marks

# Define a average marks method here
    def average_marks(self):
        return sum(self.marks) / len(self.marks)
        
```

```python
# Create 3 student objects
Student1 = Student("Kunal", 13, [70, 65, 55])
Student2 = Student("Pankaj", 7, [80, 55, 84])
Student3 = Student("siddhi", 19, [65, 90, 63])
```

```python
# Display average marks
print(Student1.name, Student1.average_marks())
print(Student2.name, Student2.average_marks())
print(Student3.name, Student3.average_marks())
```
