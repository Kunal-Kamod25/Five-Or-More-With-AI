Problem 1. Write a program that accepts marks of five subjects from the user, stores them in a list,calculates the average, and prints the corresponding grade according to your own grading scheme.

```python
marks = []

for i in range(1, 6):
    #mark = float(input(f"Enter mark for subject {i}: "))
    mark = float(input("Enter mark for subject " + str(i) + ": "))
    marks.append(mark) # i want to add one new element to the end of a list.

total = sum(marks) # Calculate the total marks
average = total / len(marks) # Calculate the average marks

if average >= 90:
    grade = "A"
elif average >= 80:
    grade = "B"
elif average >= 70:
    grade = "C"
elif average >= 60:
    grade = "D"
else:
    grade = "F"
    
print("Entered Marks:", marks)
print("Total Marks:", total)
print("Average Marks:", average)
print("Grade:", grade)
```

Problem 8. Write a program that counts the frequency of each character in a given string using a dictionary. Example: banana should produce b : 1 a : 3 n : 2

```python
text = input("Enter a string: ")

frequency = {}

for ch in text:
    if ch in frequency:
        frequency[ch] += 1
    else:
        frequency[ch] = 1

print("Character Frequency:")

for ch in frequency:
    print(ch, ":", frequency[ch])
```

Problem 9. Accept two sets from the user and display • Union • Intersection • Difference • Symmetric Difference

```python
set1 = set(map(int, input("Enter elements of Set 1: ").split()))
set2 = set(map(int, input("Enter elements of Set 2: ").split()))

union = set1 | set2
intersection = set1 & set2
difference = set1 - set2
symmetric_difference = set1 ^ set2

print("Set 1:", set1)
print("Set 2:", set2)
print("Union:", union)
print("Intersection:", intersection)
print("Difference:", difference)
print("Symmetric Difference:", symmetric_difference)
```

Problem 10. Create a simple contact book using a dictionary. The program should allow the user to • Add a contact • Search for a contact • Update a contact • Delete a contact • Display all contacts Use a menu-driven approach.

```python
contacts = {}

while True:
    print("\n..Contact Book ....")
    print("1. Add Contact")
    print("2. Search Contact")
    print("3. Update Contact")
    print("4. Display All Contacts")
    print("5. Exit")
    choice = input("Enter choice: ")
    
    if choice == '1':
        name = input("Name: ")
        number = input("Phone number: ")
        contacts[name] = number
        print("Contact added successfully.")
        
    elif choice == '2':
        name = input("Name to search: ")
        if name in contacts:
            print("Phone number:", contacts[name])
        else:
            print("Contact not found.")
            
    elif choice == '3':
        name = input("Name to update: ")
        if name in contacts:
            number = input("Enter new phone number: ")
            contacts[name] = number
            print("Contact updated successfully.")
        else:
            print("Contact not found.")
            
    elif choice == '4':
        name = input("Name to delete: ")
        if name in contacts:
            del contacts[name]
            print("Contact deleted successfully.")
        else:
            print("Contact not found.")
            
    elif choice == '5':
            if not contacts:
                print("No contacts found.")
            else:
                print("All Contacts:")
                for name, number in contacts.items():
                    print(f"{name}: {number}")
            
    elif choice == '6':
        print("Exiting...")
        break
    else:
        print("Invalid choice.")
```

Problem 13. Generate Floyd’s Triangle for a given value of n.

```python
n = int(input("Enter n: "))
counter = 1
for i in range(1, n + 1):
    row = " "
    for j in range(i):
        row = row + str(counter) + " "
        counter += 1
    print(row.strip())
```

Problem 14. Given a list of integers, remove all duplicate values while preserving the original order of the elemnts.

```python
n = int(input("Enter n: "))
number = []
for i in range(n):
    num = int(input(f"Enter number {i + 1}: "))
    number.append(num)
    
seen = set()
result = []
for num in number:
    if num not in seen:
        seen.add(num)
        result.append(num)
    print(f"List without duplicates: {result}")
```

Problem 15. Write a menu-driven program called Student Record Management System. Each student record should contain • Name • Roll Number • Course • Marks The program should allow the user to • Add a student • Search for a student • Update student information • Delete a student • Display all students • Exit Store the records using appropriate Python data structures.

```python
students = {}

while True:
    print("Student Record Management System")
    print("1. Add Student")
    print("2. Search Student")
    print("3. Update Student")
    print("4. Delete Student")
    print("5. Display All Students")
    print("6. Exit")
    choice = int(input("Enter your choice: "))
    
    if choice == 1:
        rollNo = input("Enter Roll Number: ")
        name = input("Name: ")
        course = input("Course: ")
        marks = float(input("Marks: "))
        students[rollNo] = {"Name": name, "Course": course, "Marks": marks}
        print("Student added.")
        
    elif choice == 2:
        rollNo = input("Enter Roll Number to search: ")
        if rollNo in students:
            print(students[rollNo])
        else:
            print("Student not found.")
            
    elif choice == 3:
        rollNo = input("Enter Roll Number to update: ")
        if rollNo in students:
            students[rollNo]["Name"] = input("Name: ")
            students[rollNo]["Course"] = input("Course: ")
            students[rollNo]["Marks"] = float(input("Marks: "))
            print("Student updated.")
        else:
            print("Student not found.")
            
    elif choice == 4:
        rollNo = input("Enter Roll Number to delete: ")
        if rollNo in students:
            del students[rollNo]
            print("Student deleted.")
        else:
            print("Student not found.")
            
    elif choice == 5:
        if not students:
            print("No students found.")
        for rollNo, details in students.items():
            print(f"rollNo {rollNo}: {details}")
            
    elif choice == 6:
        print("Exiting...")
        break
    else:
        print("Invalid choice.")
```
