```python
# Experiment 1: Swapping Variable
a = 10
b = 20

a, b = b, a

print(a)
print(b)

# Explanation:- Here python swap the value of a and b in single line. without using any extra variable.
```

```python
# Experiment 2: 

text = "Python Programming"

print ( text [0])
print ( text [ -1])
print ( text [0:6])
print ( text [:: -1])

# Explanation : 1) In first text[0] give first character ”P”. 
#          2) In Second text[-1] give last character ”g”.
#          3) In Third text[0:6] give character from index 0 to 5,”Python”. 
#          4) In Fourth text[::-1] reverse whole string because step is -1.

```

```python
# Experiment 3 : Append And Rem0ve

numbers = [10 , 20, 30]

numbers.append(40)
numbers.remove(20)
print(numbers)

# Explanation : 1) In the first We Append or add 40 at last in the list.
#          2) In the second we delete the 1 index number(20)
```

```python
# Experiment 4 : Slicing

text = "Programming"

print (text [0:5])
print (text [3:])
print (text [:6])
print (text [:: -1])

# Explanation : 
# 1) In the first we do slicing from index No. 0 To Index No.4. so we get output: Progr.and
# 2) In the second it will give from index No.3 to everything upto end of the list.
# 3) In the third it will give From Index 0 TO index No.5.
# 4) In the Fourth it will reverse the whole string. so we get output: gnimmargorP
```
